```python
# Import Libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from textblob import TextBlob   # understand natural meaning of sentences
from wordcloud import WordCloud # wordcloud
```

```python
# Load the dataset

data = pd.read_csv("Sentiment_Analysis.csv")

data.head(10)
```

```python
# Define the sentiment analysis function

def get_sentiment(text):

    meaning = TextBlob(text)

    if meaning.sentiment.polarity>0:
        return "Positive"

    elif meaning.sentiment.polarity<0:
        return "Negative"

    else:
        return "Neutral"
```

```python
data["Sentiment"] = data["Sentences"].apply(get_sentiment)
```

```python
data.head(20)
```

```python
data["Sentiment"].value_counts()
```

```python
# filter out all the negative reviews

negative_reviews = data[data["Sentiment"]=="Negative"]

negative_reviews.head(9)
```

```python
# Combine/Merge all negative sentence into one paragraph/string

all_negative_text = " ".join(negative_reviews["Sentences"])
```

```python
print(all_negative_text)
```

```python
# Create the word cloud for negative reviews

diagram = WordCloud(width=800, height=800, background_color="white",colormap="magma").generate(all_negative_text)
```

```python
# Plot the word cloud

plt.figure(figsize=(10,10))
plt.imshow(diagram,interpolation="bilinear")
plt.title("Most Common Words in Negative Feedbacks")
plt.axis("off")
plt.show()
```
